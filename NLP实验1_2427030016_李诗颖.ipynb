```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

en_text = "Hello world, I study python and natural language processing."
tokens = word_tokenize(en_text)
stop_words = set(stopwords.words('english'))
filtered = [w for w in tokens if w.lower() not in stop_words]
print("英文分词：", tokens)
print("过滤停用词：", filtered)

import jieba
cn_text = "今天在VSCode学习自然语言处理，中英文分词都要掌握"
print("中文分词：", jieba.lcut(cn_text))
```

```python
import jieba
from collections import Counter

text = '自然语言处理是人工智能皇冠上的明珠，Python是自然语言处理常用的编程工具。'
words = jieba.lcut(text)
print('分词结果：', words)

word_freq = Counter(words)
print('词频统计：', word_freq.most_common(10))
```
